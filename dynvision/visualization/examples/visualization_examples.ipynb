# DynVision Interactive Visualization Examples

This notebook demonstrates the interactive visualization capabilities of DynVision, focusing on:
1. Model Architecture Visualization
2. Weight Distribution Analysis
3. Temporal Dynamics Visualization

```python
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

from dynvision.models import CorNet
from dynvision.visualization.plot_model_architecture import ModelArchitectureViz, plot_temporal_flow
from dynvision.visualization.enhanced_weight_distributions import WeightDistributionViz, plot_weight_correlation_matrix

%matplotlib widget
```

## 1. Model Architecture Visualization

Visualize the network architecture with special focus on recurrent connections.

```python
# Initialize a model
model = CorNet()

# Create architecture visualizer
arch_viz = ModelArchitectureViz(model)

# Plot architecture
fig, ax = arch_viz.plot_architecture(fig_size=(15, 10))

# Print architecture summary
arch_viz.print_architecture_summary()
```

## 2. Weight Distribution Analysis

Analyze weight distributions across layers with statistical insights.

```python
# Create weight distribution visualizer
weight_viz = WeightDistributionViz(model)

# Plot enhanced weight distributions
fig, axes = weight_viz.plot_weight_distributions(split_by_type=True)

# Print weight statistics summary
weight_viz.print_weight_summary()

# Plot weight correlations
fig_corr, ax_corr = plot_weight_correlation_matrix(model)
```

## 3. Temporal Dynamics Visualization

Visualize how activations flow through the network over time.

```python
# Create sample input
input_size = (1, 3, 224, 224)  # Adjust based on your model

# Plot temporal flow
fig, ax = plot_temporal_flow(model, input_size, timesteps=10)

# Customize the plot
plt.grid(True)
plt.title('Temporal Flow of Activations')
```

## 4. Interactive Analysis

Demonstrate interactive features for detailed analysis.

```python
from ipywidgets import interact, interactive, fixed, widgets

def update_weight_plot(layer_type='all', show_stats=True):
    fig, axes = weight_viz.plot_weight_distributions(
        split_by_type=(layer_type == 'split'),
        show_stats=show_stats
    )
    plt.show()

# Create interactive widget
interact(
    update_weight_plot,
    layer_type=widgets.Dropdown(
        options=['all', 'split'],
        value='all',
        description='Layer Display:'
    ),
    show_stats=widgets.Checkbox(
        value=True,
        description='Show Statistics'
    )
)
```
